## ex. 1

$$  
y = x^Tx,  \quad x \in \mathbb{R}^N
$$

$$
\frac{dy}{dx} = 2x
$$

Покоординатно: $y = x^Tx = \sum_{i=1}^N x_i^2$, поэтому
$$
\frac{\partial y}{\partial x_k} = \frac{\partial}{\partial x_k}\sum_{i=1}^N x_i^2 = 2x_k
\quad\Longrightarrow\quad \frac{dy}{dx} = 2x .
$$

## ex. 2

$$ y = tr(AB) \quad A,B \in \mathbb{R}^{N \times N} $$

$$
\frac{dy}{dA} = B^T
$$

$$
y = tr(AB) = \sum_{i=1}^N (AB)_{ii} = \sum_{i=1}^N\sum_{j=1}^N A_{ij}B_{ji}.
$$
Элемент $A_{ij}$ входит в сумму ровно один раз, с коэффициентом $B_{ji}$:
$$
\frac{\partial y}{\partial A_{ij}} = B_{ji}
\quad\Longrightarrow\quad
\frac{dy}{dA} = B^T .
$$

## ex. 3

$$  
y = x^TAc , \quad A\in \mathbb{R}^{N \times N}, x\in \mathbb{R}^{N}, c\in \mathbb{R}^{N}
$$

$$
\frac{dy}{dx} = Ac
$$

Обозначим $b = Ac \in \mathbb{R}^N$ (не зависит от $x$). Тогда $y = x^Tb = \sum_i x_i b_i$ и
$$
\frac{\partial y}{\partial x_k} = b_k \quad\Longrightarrow\quad \frac{dy}{dx} = b = Ac .
$$

$$
\frac{dy}{dA} = xc^T
$$

Hint for the latter (one of the ways): use *ex. 2* result and the fact
$$
tr(ABC) = tr (CAB)
$$

$y$ — скаляр, поэтому $y = tr(y) = tr(x^TAc)$. По цикличности следа $tr(ABC) = tr(CAB)$:
$$
y = tr(x^T A c) = tr(c\,x^T A) = tr\big(A\,(c\,x^T)\big).
$$
Аналогично *ex. 2* с $B = c\,x^T$:
$$
\frac{dy}{dA} = B^T = (c\,x^T)^T = x\,c^T .
$$

## ex. 4

Classic matrix factorization example. Given matrix $X$ you need to find $A$, $S$ to approximate $X$. This can be done by simple gradient descent iteratively alternating $A$ and $S$ updates.
$$
J = || X - AS ||_F^2  , \quad A\in \mathbb{R}^{N \times R} , \quad S\in \mathbb{R}^{R \times M}
$$
$$
\frac{dJ}{dS} = -2A^T(X - AS) = 2A^T(AS - X)
$$

You may use one of the following approaches:

#### First approach
Using ex.2 and the fact:
$$
|| X ||_F^2 = tr(XX^T)
$$
it is easy to derive gradients (you can find it in one of the refs).

**Решение:**

Здесь $X \in \mathbb{R}^{N\times M}$. Раскроем норму через след:
$$
J = tr\big((X - AS)(X - AS)^T\big) = tr(XX^T) - tr(XS^TA^T) - tr(ASX^T) + tr(ASS^TA^T).
$$
Так как $tr(M) = tr(M^T)$, два средних слагаемых равны: $tr(XS^TA^T) = tr(ASX^T)$, поэтому
$$
J = tr(XX^T) - 2\,tr(ASX^T) + tr(ASS^TA^T).
$$

1. $tr(XX^T)$ от $S$ не зависит, производная $0$.

2. $tr(ASX^T) = tr(S\,X^TA)$ (цикличность). По *ex. 2* с $B = X^TA$:
$$
\frac{d}{dS}\,tr(S\,X^TA) = (X^TA)^T = A^TX .
$$

3. $tr(ASS^TA^T) = tr(S S^T A^TA)$. Возьмём дифференциал, обозначив $C = A^TA$ (симметричная):
$$
d\,tr(SS^TC) = tr(dS\,S^TC) + tr(S\,dS^T C) = tr(dS\,S^TC) + tr(C^T dS\,S^T) = 2\,tr(dS\,S^TC),
$$
(во втором слагаемом транспонировали след и применили цикличность). Снова по *ex. 2* с $B = S^TC$:
$$
\frac{d}{dS}\,tr(ASS^TA^T) = 2\,(S^TA^TA)^T = 2A^TAS .
$$

Итого:
$$
\boxed{\frac{dJ}{dS} = -2A^TX + 2A^TAS = -2A^T(X - AS)}
$$

#### Second approach
You can use *slightly different techniques* if they suits you. Take a look at this derivation:
<img src="https://github.com/girafe-ai/ml-course/blob/23f_basic/homeworks/lab01_ml_pipeline/grad.png?raw=1">
(excerpt from [Handbook of blind source separation, Jutten, page 517](https://books.google.ru/books?id=PTbj03bYH6kC&printsec=frontcover&dq=Handbook+of+Blind+Source+Separation&hl=en&sa=X&ved=0ahUKEwi-q_apiJDLAhULvXIKHVXJDWcQ6AEIHDAA#v=onepage&q=Handbook%20of%20Blind%20Source%20Separation&f=false), open for better picture).

#### Third approach
And finally we can use chain rule!
let $ F = AS $

**Find**
$$
\frac{dJ}{dF} = -2(X - F) = 2(F - X)
$$
and
$$
\frac{dF}{dS} = I_M \otimes A
$$
(the shape should be $ NM \times RM )$.

Now it is easy do get desired gradients:
$$
\frac{dJ}{dS} = 2A^T(AS - X)
$$

**Решение:**

*Шаг 1.* $J = \|X - F\|_F^2 = \sum_{i,j}(X_{ij} - F_{ij})^2$, поэтому
$$
\frac{\partial J}{\partial F_{ij}} = -2(X_{ij} - F_{ij}) \quad\Longrightarrow\quad \frac{dJ}{dF} = -2(X - F) \in \mathbb{R}^{N\times M}.
$$

*Шаг 2.* $F_{ij} = \sum_{k=1}^R A_{ik}S_{kj}$, значит
$$
\frac{\partial F_{ij}}{\partial S_{kl}} = A_{ik}\,\delta_{jl}.
$$
Чтобы записать эту 4-индексную производную матрицей, векторизуем матрицы по столбцам ($vec$). Столбец $j$ матрицы $F$ равен $A$, умноженной на столбец $j$ матрицы $S$: $F_{:,j} = A\,S_{:,j}$. Поэтому
$$
vec(F) = \begin{pmatrix} A & & \\ & \ddots & \\ & & A \end{pmatrix} vec(S) = (I_M \otimes A)\,vec(S),
\qquad
\frac{d\,vec(F)}{d\,vec(S)} = I_M \otimes A \in \mathbb{R}^{NM \times RM}.
$$
Это блочно-диагональная матрица из $M$ копий $A$, ровно требуемой формы $NM \times RM$.

*Шаг 3 (цепное правило).*
$$
\frac{dJ}{d\,vec(S)} = \left(\frac{d\,vec(F)}{d\,vec(S)}\right)^T \frac{dJ}{d\,vec(F)} = (I_M \otimes A^T)\,vec\big(-2(X - F)\big) = vec\big(-2A^T(X - AS)\big),
$$
т.к. $(I_M\otimes A^T)\,vec(Y) = vec(A^TY)$ (каждый столбец $Y$ умножается на $A^T$). Возвращаясь к матричной форме:
$$
\frac{dJ}{dS} = -2A^T(X - AS) = 2A^T(AS - X),
$$

### Question 2

We can also use other distance metrics such as L1 distance.
For pixel values $p_{ij}^{(k)}$ at location $(i,j)$ of some image $I_k$,

the mean $\mu$ across all pixels over all images is $$\mu=\frac{1}{nhw}\sum_{k=1}^n\sum_{i=1}^{h}\sum_{j=1}^{w}p_{ij}^{(k)}$$
And the pixel-wise mean $\mu_{ij}$ across all images is
$$\mu_{ij}=\frac{1}{n}\sum_{k=1}^np_{ij}^{(k)}.$$
The general standard deviation $\sigma$ and pixel-wise standard deviation $\sigma_{ij}$ is defined similarly.

Which of the following preprocessing steps will not change the performance of a Nearest Neighbor classifier that uses L1 distance? Select all that apply.
1. Subtracting the mean $\mu$ ($\tilde{p}_{ij}^{(k)}=p_{ij}^{(k)}-\mu$.)
2. Subtracting the per pixel mean $\mu_{ij}$  ($\tilde{p}_{ij}^{(k)}=p_{ij}^{(k)}-\mu_{ij}$.)
3. Subtracting the mean $\mu$ and dividing by the standard deviation $\sigma$.
4. Subtracting the pixel-wise mean $\mu_{ij}$ and dividing by the pixel-wise standard deviation $\sigma_{ij}$.
5. Rotating the coordinate axes of the data.

*Your Answer:*

1, 2, 3

*Your Explanation:*

Классификатор ближайшего соседа зависит только от *порядка* расстояний: если преобразование меняет все расстояния одинаково монотонно, ответы не меняются. L1-расстояние между изображениями $I_k$ и $I_l$: $d_1 = \sum_{i,j}|p_{ij}^{(k)} - p_{ij}^{(l)}|$.

1. **Вычитание $\mu$** — сдвиг всех пикселей на одну константу, она сокращается в разности: $|(p^{(k)}_{ij} - \mu) - (p^{(l)}_{ij} - \mu)| = |p^{(k)}_{ij} - p^{(l)}_{ij}|$. Расстояния не меняются - ок
2. **Вычитание $\mu_{ij}$** — сдвиг в каждой координате свой, но одинаковый для всех изображений, поэтому в разности тоже сокращается. Расстояния не меняются - ок
3. **Вычитание $\mu$ и деление на $\sigma$** — сдвиг сокращается, а деление на одну и ту же константу $\sigma>0$ умножает все расстояния на $1/\sigma$. Порядок соседей сохраняется - ок
4. **Вычитание $\mu_{ij}$ и деление на $\sigma_{ij}$** — каждая координата масштабируется *своим* коэффициентом: $d_1 = \sum_{i,j}\frac{1}{\sigma_{ij}}|p_{ij}^{(k)} - p_{ij}^{(l)}|$. Это взвешенная L1-метрика, веса пикселей меняются, и ближайший сосед может стать другим. Пример в 2D: $x=(0,0)$, $a=(2,0)$, $b=(0,3)$; до масштабирования $d(x,a)=2<d(x,b)=3$, а после деления второй координаты на 3 получаем $d(x,b)=1<2$ - не ок
5. **Поворот осей** — L1-норма не инвариантна относительно поворотов. Пример в 2D: $x=(0,0)$, $a=(1,0)$, $b=(0.8,0.8)$. До поворота $d_1(x,a)=1 < d_1(x,b)=1.6$, ближайший — $a$. После поворота на $45^\circ$: $a\to(0.71,\,0.71)$, $d_1=1.41$; $b\to(0,\,1.13)$, $d_1=1.13$ — теперь ближайший $b$. Порядок соседей изменился - не ок

## Question 3

Which of the following statements about $k$-Nearest Neighbor ($k$-NN) are true in a classification setting, and for all $k$? Select all that apply.
1. The decision boundary (hyperplane between classes in feature space) of the k-NN classifier is linear.
2. The training error of a 1-NN will always be lower than that of 5-NN.
3. The test error of a 1-NN will always be lower than that of a 5-NN.
4. The time needed to classify a test example with the k-NN classifier grows with the size of the training set.
5. None of the above.

*Your Answer:*

2, 4

*Your Explanation:*

1. **Неверно.** Граница решений k-NN — кусочно-линейная (для 1-NN это границы ячеек диаграммы Вороного между точками разных классов) и в целом сильно нелинейная, может быть даже несвязной. Линейной она не является.
2. **Верно** (в смысле "не выше"). При оценке на обучающей выборке ближайший сосед каждой обучающей точки — она сама (расстояние 0), поэтому 1-NN имеет нулевую ошибку на train (если нет совпадающих точек с разными метками). У 5-NN к самой точке добавляются ещё 4 соседа, которые могут перевесить её метку, так что его ошибка на train $\geq 0$ и, как правило, больше. Строгое равенство возможно лишь в вырожденном случае, когда и 5-NN не ошибается ни на одной точке.
3. **Неверно.** 1-NN сильно переобучается и чувствителен к шуму в метках; на тесте 5-NN часто работает лучше за счёт усреднения по нескольким соседям. Оптимальное $k$ подбирают на валидации, и это не обязательно $k=1$.
4. **Верно.** k-NN — "ленивый" алгоритм: обучение — это запоминание данных, а для классификации нужно посчитать расстояния от тестового объекта до всех $n$ обучающих, что занимает $O(n\cdot D)$. Время предсказания растёт с размером обучающей выборки (структуры вроде KD-дерева уменьшают константу/степень, но не убирают зависимость от $n$).
